# Interacting with the Environment

## Computer Vision
The area of computer vision has a close relationship with deep learning algorithms and models, which have proven very effective to solve computer vision problems like:
- **Image processing:** intellegently enhance images and video
    - Photografic filters, etc
- **Object detection:** identify subjects in images/video for surveillance, logistics, etc
    - Surveillance, package tracking, etc
- **Motion Analysis:** extract motion information like speed and direction of objects
- **Image and video generation:** create realistic visual data from human text

## Natural Language processing (NLP)
The most popular area today. The range of problems NLP can address is growing fast:
- **Text-based**
    - **Sentiment analysis:** extract positive and negative feelings in text, e.g. customer reviews
    - **Question Answering** (chatbots)
    - **Text summarization** 
- **Speech-based**
    - Text-to-speech
    - Speech-to-text

## Robotics
Area of AI where the interaction with the physical environment is nearly omnipresent.

- Combine computer vision and NLP in a wide number of applications.
- Divided into:
    - **Sensing and perception:** collecting data or perceiving signals from the environment
        - This is important not only to collect outside data, for example, aerial images captured by drones, but also to make decisions and execute other actions like mobility
    - **Mobility:** moving the environment guided by perceptions of surroundings
        - For example, a mobile robot may decide to stop moving and change the course of action when its proximity sensors encounter an obstacle in the surroundings
    - **Manipulation:** the robot modifies the environment
        - For example, industrial robots that assemble vehicles by grabbing, releasing, and assembling pieces together
    - **Human-robot interaction:** e.g. conversational robots endowed with NLP
        - Robots that speak and listen to humans
