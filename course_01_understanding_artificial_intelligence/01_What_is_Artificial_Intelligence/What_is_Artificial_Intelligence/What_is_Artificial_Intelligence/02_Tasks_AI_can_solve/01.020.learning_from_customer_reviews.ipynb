# Learning from customer reviews

An e-commerce brand has realized they have millions of customers' text reviews of products in their sample catalog. Now they have designed a solution that analyzed these countless pieces of text to find out aspects of products that customers tend to like or dislike the most, in order to design new products in the future.

Which area of NLP is most important in this scenario?

## Possible Answers
1. Text summarization
2. Sentiment analysis
3. Text classification
4. Translation

## Correct Answer
2. Sentiment analysis

Perfect! Sentiment analysis is definitely the solution behind this problem of inferring what customers praise and dislike about every product in the catalog.
