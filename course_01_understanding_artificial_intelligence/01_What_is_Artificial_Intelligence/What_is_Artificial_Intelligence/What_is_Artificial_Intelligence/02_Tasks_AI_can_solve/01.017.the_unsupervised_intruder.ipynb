# The Unsupervised Intruder

Both supervised and unsupervised learning encompasses a wealth of techniques to solve various problems and tasks.

Now it's time for you to distinguish between problems solvable with supervised learning and those addressed by unsupervised learning.

The following list of supervised learning uses has an "intruder": which of these problems should actually be solved under an unsupervised setting?

## Possible Answers
1. Forecasting the daily petrol price per liter along the next week, given last year's daily prices
2. Given a collection of dog pictures, identifying the dog breed each one belongs to
3. Detecting unusual peaks in the EUR-GBP currency exchange rate

## Correct Answer
3. Detecting unusual peaks in the EUR-GBP currency exchange rate

Perfect! You found the intruder: detecting something abnormal or unusual in the data is an unsupervised learning task, concretely anomaly detection.
