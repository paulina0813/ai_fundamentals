Data acquisitionis of these AI funcitons, and so is learning and reasoning from data.

Once they get the data, AI systems and their underlying algorithms must learn or apply some reasoning with those data to yield outputs in the form of decisions, actions, or insights.

Machine Learning and Deep Learning have a lot to say in this part. 

In addition, some types of AI systems, for example robots or computer vision systems require interacting with the physical environment to collect data inputs or perform actions based on learning and reasoning outputs.

As we may expect, AI areas play a particular role in these functions

### Data Acquisition

- There are different approaches to data acquisition

    #### Data Acquisition: sensing the environment
    - Collect outside sensory information like sounds, images, etc. through sensors, emulating the five human senses
    - Turn perceptions into data
    - The perception of taste or smell by AI is still not too explored, but the rest of the senses are already commonly used
    - Occurs in:
        - **NLP and audio:** capturing speech and sounds
            - Speech and sounds can be captured by NLP and audio tasks like song classification
        - **Computer Vision:** satellite images, fingerprint, etc
            - They acquire visual stimuli as large as satellite images or as small as fingerprints
        - **Robotics and sensors:** temperature, touch, motion, gravity, etc
            - In robotics and sensor networks, like IoT systems, sensing the environment is particularly common, be it perceiving temperature, touch, motion, and gravity, to name a few.
    
    #### Data Acquisition: datasets
    Whether data comes from sensing the outside environment or not, AI systems deal with sets of input data called datasets.
    - **Dataset:** a collection of data, namely a set of data samples, observations, or instances of a certain type of data.
        - **Structured:** tabular format, spreadsheets
        - **Unstructured:** images, audio, videos, text, ...
    Depending on the nature of the data and the problem to address, datasets might be manually created by a human.

    This is typically the case for smaller tabular datasets such as certain database tables.

    However, a vast majority of AI applications nowadays rely on datasets built automatically.

    For example, in an e-commerce portal, every customer purchase of a product triggers the creation of new instances in customer and product-related datasets

    And finally, the mechanisms seen earlier to sense the environmemt are another way to create datasets, normally as a particular case of automatically collected data
