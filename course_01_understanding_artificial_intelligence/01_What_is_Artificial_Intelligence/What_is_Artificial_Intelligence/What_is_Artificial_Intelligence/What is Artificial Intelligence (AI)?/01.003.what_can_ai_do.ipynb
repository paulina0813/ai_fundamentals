# What can AI do?

AI is capable of making predictions, recognizing patterns in data, optimizing complex processes, and automating repetitive tasks.

This is not a complete list but it does represent some of AI's most impactful applications.

## Machine Learning
Allows systems to learn from past data and make decisions.

Two key tasks of machine learning are:

|Predictions|Inferences|
|---|---|
|Help forecast future events, like high precision weather forecasting|Focus on determining an outcome based on existing data, such as recommending a book you might enjoy based on your past preferces|

## AI
AI excels at pattern recognition, which involves analyzing data to identify trends and make informed decisions.

AI powered clustering techniques help businesses segment customers into groups with similar behaviors, making it easier to tailor marketing strategies

In security and finance, anomaly detection sports unusual transactions or activities that might indicate fraud. 

Generative AI uses patterns in existing data to create new content, such as text, images, or music.

AI is a powerful tool for solving optimization problems. It helps delivery services determine the fastest routes, supports energy grids in operating efficiently, and enables travel companies to set dynamic pricing strategies that maximize revenue.

Retailers and brands also use AI to plan discount campaigns that drive sales by analyzing customer behavior.

Another key AI application is automation, where machines follow predefined rules to complete tasks independently.

Although automation isn't AI, since it doesn't involve learning or reasoning, AI enhances automation by improving efficiency.

AI can quickly analyze and classify large batches of documents and images, screen job applications, and manage warehouse logistics with robotic systems

## AI Limitations
Conversational AI, like chatbots, can answer questions but struggles with social nuances such as emotional intelligence and empathy.

AI powered recommendation systems may have difficulty handling new and unfamiliar situations, like suggesting a brand new product with no prior customer data.

Bias is another challenge, as AI systems can produce unfair predictions if trained on biased data, favoring certain groups over others based on factors like gender, age, or ethnicity.

AI is only as good as the data it learns from. Without high-quality data even the most advanced AI models become unreliable.
