# How does AI learn?

AI can do amazing things, from recognizing faces to predicting the weather. One of its most important skills is learning from data to improve over time.

Which AI field specializes in teaching computers to learn from data, recognize patterns, and make predictions - without needing to be explicitly programmed for every task?

## Possible Answers
1. Machine Learning
2. Natural Language Processing (NLP)
3. Computer Vision
4. Robotics

## Correct Answer
1. Machine Learning
