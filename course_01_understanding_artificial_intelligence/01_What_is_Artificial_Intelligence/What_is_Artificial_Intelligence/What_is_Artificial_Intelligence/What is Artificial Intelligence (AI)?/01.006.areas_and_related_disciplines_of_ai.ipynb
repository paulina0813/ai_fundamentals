# Areas and related disciplines of AI

## Machine Learning
Teaching computers to:
- Learn from data
- Recognize patterns
- Make predictions and decisions

### Deep Learning
- Branch of machine learning
- Uses neural networks to tackle big challenges
- Recognize images
- Play games

AI also needs a wat to process and apply logic, which is where knowledge representation and reasoning come in. This area helps AI systems understand informations and make smart decisions.

Robotics brings AI into the real world, allowing machines to interact with their surroundings, while computer vision give AI the ability to "see" and understand images and objects

## Computer Vision
- Allows computers to "see" images and objects

## Natural Language Processing
- NLP
- Enables AI to understand and generate human languages - text and speech

The most advanced AI systems don't work in isolation. Many combine different areas to combine smarter, more capable applications.

## Examples
- An e-commerce recommendation system doesn't just guess what you might like - it learns from your shopping habits using Machine Learning
- In a smary warehouse, AI-powered robots navigate the space using Computer Vision while communicating and reasoning to optimize storage and deliveries.
- In healthcare, Deep Learning and Computer Vision team up to analyze X-rays and assist doctors in diagnosing coditions
- Voice assistants use NLP and Deep Learning to understand speech and interact with users naturallt

## Data Science
One of AI's closest allies
- Focuses on extracting insights from data
- AI and data science both rely on mathematics and statistics, using concepts like probability and linear algebra to drive their models
