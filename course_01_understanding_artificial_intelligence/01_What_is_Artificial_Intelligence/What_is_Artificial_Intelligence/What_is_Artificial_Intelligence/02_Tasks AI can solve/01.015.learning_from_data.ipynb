# Learning from Data

## Machine Learning (ML)
- **Machine Learning:** consists of learning from data to identify patterns
    - Supervised learning: for predictive tasks like classification and regression
    - Unsupervised learning: specializes in other pattern recognition tasks including clustering and anomaly detection
    - Reinforcement Learning: investigates agents that learn by themselves from past experiences
    - Deep Learning: use so-called neural networks to solve much more complex variants of these ML problems

    !["Machine Learning"](<ml.png>)

    ### Supervised Learning: classification
    - **Classification:** consists of assigning the right category or class to new data obersvations whose class is unknown. 
        - **Binary Classification:** two classes, e.g. positive/negative, male/female, etc.

        
        !["Binary Classification"](<binary_classification.png>)


        - **Multi-class classification:** several mutually exclusive classes, e.g. multiple species


        !["Multi-class Classification"](<multi_class_classification.png>)


        **Supervised Learning:** _Data annotation_ (getting labelled observations with known class a priori) needed to learn/train a model capable of making inference
    
    ### Supervised Learning: regression and forecasting
    - **Regression:** is the numerical counterpart of classification, where the target value to predict is quantitative - a number
        - Predicting a house value based on some attributes is an example of a problem solvable via regression


        !["Regression"](<regression.png>)


    - **Time series forecasting:** the goal is to predict the future values of out tardet, based on observations from the past
        - For instance, forecasting the number of daily bus passengers for the next momth, based on historical data

        !["Forecasting"](<forecasting.png>)
    
    ### Unsupervised and reinforcement learning
    Tries to learn properties or patterns behind the data, without assigning them to a predefined label.

    - **Clustering:** find subgroups of data with similar characteristics.
    - **Anomaly detection:** detecting abnormal data observations e.g. unusual card transactions
    - **Association rule discovery:** find common co-occurrences of items in transaction data
    - **Reinforcement learning:** learn by experience (trial and error) to master a complex task


    !["Unsupervised and Reinforcement Learning"](<unsupervised_and_reinforcement_learning.png>)


    ### How about Deep Learning?
    - Highly sophisticated models based on deep neural networks: solve very challenging tasks where classical ML models become limited
    - It teaches computers to learn from data as a human brain would do
    - Deep Learning models look for patterns in pictures, text, videos, sounds, and other data by using so-called deep neural networks
    - These powerful models require huge amounts of data to be trained.
    - Thanks to technological advancements in computing and storing data, deep learning has gone a lot further in recent years
    - It can now perform impressive tasks like recognizing objects in videos, translating and summarizing text, and even generating realistic conversations, images, and songs
