# Data sharing for AI progress

The availability of high-quality data is a central aspect to foster AI progress, for instance:

- By giving research communities access to datasets to develop new models and techniques.
- By developing new AI-driven products and services.

There are different practical considerations in data sharing. Some of them are good practices from an ethics and privacy viewpoint, while others are rather deemed as poor data sharing practices.
