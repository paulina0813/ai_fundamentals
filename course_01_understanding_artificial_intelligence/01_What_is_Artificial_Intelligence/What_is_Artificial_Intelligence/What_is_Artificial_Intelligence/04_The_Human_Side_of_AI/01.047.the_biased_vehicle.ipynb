# The biased vehicle

An autonomous car powered by AI is currently in the experimental trial stage. The tests take place in a middle-latitude region with a continental climate (warm summers, cold winters, and occasional rain).

The evaluation committee is proposing four different initiatives for comprehensive testing of the vehicle, both in a simulated environment and on actual roads.

One of these initiatives might lead to a biased AI system if it were implemented, such that the vehicle cannot behave properly or make informed decisions under certain conditions.

Identify which one might be the "problematic" initiative?

## Possible Answers
1. Simulate various driving scenarios, including urban, suburban, and rural settings with varying traffic levels.
2. Implement multiple safety measures, including redundant sensors and emergency stop procedures.
3. Evaluate and improve the AI system through rigorous real-world testing in edge scenarios like low visibility bends and turns.
4. Real-road trialing of the vehicle, fueled by data collected on the target roads during the summer months.

## Correct Answers
4. Real-road trialing of the vehicle, fueled by data collected on the target roads during the summer months.

Bingo! This initiative raised a kind of data bias called 'sampling bias': the data used to train or test the AI system is not representative of all the real-world conditions it may encounter. If the car is only trained upon road data collected in summer, a cold and snowy winter may later pose an unknown risk.
