# Biased or unbiased behavior?

BalanceBit, a fictional AI ethics organization, created a large language model (LLM) and chatbot to highlight potential sources of ethical concerns. Their chatbot, BalanceBanter, provides an interactive experience to discuss common issues around utilizing AI for many different scenarios.

You have been hired to assess BalanceBanter's responses to common prompts.

Which two scenarios receive a response that demonstrates using AI ethically.

## Possible Answers
1. Screening resumes
2. Smart agriculture
3. Hotel tourism
4. Social media moderation

## Correct answers
3. Hotel tourism
4. Social media moderation
