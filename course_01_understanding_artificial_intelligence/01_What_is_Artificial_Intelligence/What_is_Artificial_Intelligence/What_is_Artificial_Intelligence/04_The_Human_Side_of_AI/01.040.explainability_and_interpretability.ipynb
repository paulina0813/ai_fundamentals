# Explainability and interpretability

- **Explainability:** humans' ability to access and understand AI outputs, e.g. predictions, decisions
    - An example of an explainable AI model is an an ML classifier for penguin species in which you can easily find out why a certain penguin data obersvation has been classified as a Gentoo penguin


    !["Explainability Example"](<explainability_example.png>)


- **Interpretability:** understand AI systems' internal processes: algorithm, model, data workflow


- Explainability focuses on explaining outputs from inputs and interpretability aims at unveiling the internal AI process in between


    !["Interpretability Example"](<interpretability_example.png>)


## White-box vs black-box AI systems
Based on their degree of interpretability or transparency, AI systems - particularly ML/DL models - can be classified into white-box and black-box models

- **White-box:** transparent and easily interpretable models/systems
    - This is usually the case of classical and relatively simpler ML models like linear regression and decision trees. The last of which easily shows how a given combination of attributes' values in a new data observation leadss to a specific predicted output, as yoy can see in this example of penguin classification


    !["White-box Systems"](<white_box_systems.png>)

- **Black-box:** higher complexity, little or no degree of understandability
    - Although in practice sophistication often relates to more effective and powerful models, effectiveness normally comes at the cost of sacrificing interpretability
    - The best example is deep learning models with lots of layers of neurons, connections, and computational processes behind inferences, making it hard to understand why a particular prediction has been made


## Basic Explainable AI (XAI) tools
Making black-box models more transparent is the main subject of study in the field of Explainable AI (XAI), which encompasses different methods and tools 

- **XAI:** methods and tools to increase AI systems and models' transparency and explainability
- **Model introspection:** examining internal model parameters to understand decisions
- **Model documentation:** shareable architecture and design considerations
- **Model visualization:** human-friendly representation of data features and model outputs 


    !["XAI"](<xai.png>)



    ### XAI tools: feature importance
    - **Feature importance:** impact or contribution of features (predictors) in model outputs
        - Understand how data-driven models (ML/DL) make decisions
        - Detect and mitigate issues, e.g. biases
        - Impact on model performance if a feature were removed

    - **SHAP (SHapley Additive exPlanations)**
        - Feature importance visualizations toolbox
        - SHAP is a powerful visualization toolbox, not only capable of describing the importance of feature for a model as a whole, like in this example of a regression model to estimate admission decisions based on different academic scores

        
        !["SHAP Example 01"](<shap_example_01.png>)


        - In numerical predictive problems like regression, we can also investigate the positive and negative relationship between predictor features and the target output

        
        !["SHAP Example 02"](<shap_example_02.png>)


        - We can even get explanations for a single prediction on data obersvation, which is often the most illustrative way to convince your audience about your model's effectiveness, making it possible to explain how the attribute values in a given a observation contributed stringly or waekly, positively or negatively, to the resulting prediction value

        
        !["SHAP Example 03"](<shap_example_03.png>)


    ### Practical Implications of XAI

    - **Algorithmic Transparency:**
        - How algorithms process data and make decisions
    
    - **Local and global interpretability:**
        - Understand system behavior for a specific prediction 
        - Understand system overall behavior on a dataset or problem
    
    - **Ethical considerations:**
        - XAI to address ethical AI concerns: biases, discrimination, compliance, etc

    - **Human-AI collaboration:**
        - Reliable collaboration based on trust and feedback
