# Social challenges: ethics, fairness and privacy

## Responsible AI
- **Responsible AI:** Ethical and accountable development and use of AI systems, with regard to societal impact. It covers several aspects:
    - Ethics and fairness: fundamental to ensure AI systems adhere to ethical principles and mitigate fairness issues like biases
    - Human-centered design: places users' experience at the center of AI systems development
    - Privacy: safeguards the security of personal or sensitive data used by AI systems
    - Accountability: consists in establishing clear guidelines for AI system governance
    - Transparency: aims at making these systems explainable and interpretable
    - Sustainability: responsible AI recognizes the broader sustainable impact of AI in society and the environment

    ### Ethics and fairness
    - **AI ethics:** adhere to ethical guidelines and principles
        - Fairness -> bias has been a central topic to solve: preventing AI outputs that may cause discrimination or unfair treatment towards individuals or groups. Biases can originate from 3 soueces:
            - Data -> happens when a dataset does not meaningfully cover all possible cases, individuals, or groups in the target domain
            - Algorithms -> occurs when an algorithmic design is such that it produces more favorable outputs for certain groups than others
            - Decisions -> arise from AI system outputs that are consistently unfair, they ar einfluenced by algorithmic biases, but also by other factors such as context
        - Transparency
        - Privacy
        - Accountability
        - Liability for AI decisions

            #### Bias in AI systems: examples
            ##### Screening job resumes: data bias
            - An AI system for screening job resumes might unintentionally favor male applicants if it has been trained on historical data where past successful candidates were predominantly men
            - This eventually leads to unfair treatment of female candidates
            - Ensuring an active collection of resume data from all underrepresented group is an approach to mitigate this problem, as well as incorporating bias-correction algorithms that adjust the AI system.

            ##### E-commerce recommendations - algorithmic bias
            - Here, popular or highly purchased products might be overly promoted by the algorithm, limiting exposure to diverse options for users and disregarding other less popular products some users may like, therevy discriminating brands or sellers in the marketplace
            - This biased can be addressed by introducing techniques and metrics to promote diversity and fairness in the design and evaluation of the system

    ### Data Privacy in AI systems
    - **Data privacy:** safeguarding sensitive or personal information from unauthorized access and misuse
    - It requires implementing robust measures like fortifying data encryption protocols, anonymizin sensitive data such as contact information or ethnicity, ensuring secure data storage and sharing practices, and adhering key regulations like EU's GDPR or California's CCPA
    - By embracing these fundamentakl principles, we can mitigate risks such as data breaches and prevent discriminatory AI decisions
