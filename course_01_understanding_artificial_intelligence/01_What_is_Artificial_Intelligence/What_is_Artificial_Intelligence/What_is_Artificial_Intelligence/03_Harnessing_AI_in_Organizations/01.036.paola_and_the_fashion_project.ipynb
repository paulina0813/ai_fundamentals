# Paola and the fashion project

Recommender systems are a common AI application leveraged by numerous organizations in the retail, e-commerce, and entertainment domains.

Paola, an IT manager at a fashion retail company, leads a team constructing personalized recommendation engine that suggests fashion products to customers based on their purchase history.

This is the very first AI project Paola's company is kicking off.
