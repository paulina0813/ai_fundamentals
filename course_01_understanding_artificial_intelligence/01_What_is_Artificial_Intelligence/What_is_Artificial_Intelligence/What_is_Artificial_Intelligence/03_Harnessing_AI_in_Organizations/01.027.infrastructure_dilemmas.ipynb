# Infrastructure dilemmas

**Viewcorp**, a company specializing in AI-driven solutions for surveillance, is trying to **understand the differences between cloud-based and on-premises AI infrastructure.** They asked four advisors their opinion about these two modalities, but only one of them provided a reliable and truthful opinion!

Which of these statements best describes the key difference between cloud-based infrastructure and on-premises AI infrastructure?

## Possible Answers
1. On-premises infrastructure provides faster data processing capabilities because of its powerful and easily scalable infrastructure.
2. Cloud-based infrastructure ensures higher data security and privacy compared to on-premises infrastructure.
3. On-premises infrastructure offers a complete catalog of pre-built AI models and frameworks compared to cloud-based infrastructure.
4. Cloud-based infrastructure offers flexible resource provisioning and scalability, while on-premises infrastructure demands a fixed hardware investment.


## Correct Answer
4. Cloud-based infrastructure offers flexible resource provisioning and scalability, while on-premises infrastructure demands a fixed hardware investment.
