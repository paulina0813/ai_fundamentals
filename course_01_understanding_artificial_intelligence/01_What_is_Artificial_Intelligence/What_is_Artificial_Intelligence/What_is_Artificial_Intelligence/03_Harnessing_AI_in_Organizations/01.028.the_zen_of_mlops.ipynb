# The "zen" of MLOps

**Zen-wanders** provides immersive experiences for customers to explore the heritage, traditions, and flavors of Eastern Asian cultures from the comfort of their homes. They use **AI algorithms** that activate real-time **recommendations** for cuisine, art, music, etc., based on user's preferences.

An **MLOps lifecycle** underpins their solution, encompassing these stages:

1. Planning
2. Data preparation
3. Model training and evaluation
4. Deployment
5. Monitoring and maintenance
