# Performance stories

Below are five stories describing the performance of different kinds of AI systems and problems:

(A) A **regression solution to predict the price** (in thousand dollars) **of houses for sale,** based on their size, location, and other features.

(B) An **e-commerce recommender engine** that takes the recent purchase history of a customer and generates a short list of suggested products for them to buy next.

(C) A **cat species classification model** with a comparison between the predicted class and actual class label for a validation set of 200 cat images, 190 of which have been correctly classified.

(D) A **spam e-mail classifier,** with a weekly evolution of its performance that shows a stable declining trend over time.

(E) A **Proof-of-Concept Generative AI system** that generates realistic portrait images of people and has been trained with millions of portrait images of people from only one specific race or ethnicity.
