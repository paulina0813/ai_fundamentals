# Unmasking a hotel booking AI system

A hotel booking site with millions of customers has deployed an AI system to assign personalized hotel prices in real-time when a customer makes a hotel search, taking into account the demand rates, hotel characteristics such as amenities and proximity to points of interest, and the customer's booking history with the site.

Select which two of the following are elements of this hotel pricing AI system.

## Possible Answers
1. A pricing algorithm that maps different data inputs into an optimized hotel price for the customer
2. An algorithm that provides the customer with a list of five personalized hotel suggestions for the customer
3. A hardware cluster of server computers
4. Customer pictures of hotel facilities

## Correct Answers
1. A pricing algorithm that maps different data inputs into an optimized hotel price for the customer
3. A hardware cluster of server computers
